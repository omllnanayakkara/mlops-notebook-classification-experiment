```python
# import libraries
import os
from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential
from azure.ai.ml.entities import Workspace, ComputeInstance, AzureBlobDatastore, Data, Environment, AccountKeyConfiguration
from azure.ai.ml.constants import AssetTypes, InputOutputModes

subscription_id = os.environ.get("AZURE_SUBSCRIPTION_ID")
resource_group = os.environ.get("AZURE_RESOURCE_GROUP")

# connect azureml with default tenant-level scope
mlClient = MLClient(
    credential=DefaultAzureCredential(),
    subscription_id=subscription_id,
    resource_group_name=resource_group
)
```

```python
# create workspace
workspace = Workspace(
    name="myworkspace",
    location="eastus2",
    enable_data_isolation=True,
    resource_group=resource_group
)

mlClient.workspaces.begin_create(workspace).result()
```

```python
# connect to azureml workspace
mlClient = MLClient(
    credential=DefaultAzureCredential(),
    subscription_id=subscription_id,
    resource_group_name=resource_group,
    workspace_name="myworkspace"
)
```

```python
# create compute instance
compute_target = ComputeInstance(
    name="mytargetcomputeinstance",
    size="STANDARD_DS11_V2",
    idle_time_before_shutdown_minutes=15
)

mlClient.compute.begin_create_or_update(compute_target).result()
```

```python
# create custom environment
environment = Environment(
    name="my-env",
    version="1",
    conda_file="./conda.yml",
    image="mcr.microsoft.com/azureml/openmpi4.1.0-ubuntu20.04"
)

mlClient.environments.create_or_update(environment)
```

```python
# create datastore
datastore = AzureBlobDatastore(
    name="myblobdatastore",
    account_name="myblobstorage26",
    container_name="mycontainer",
    credentials=AccountKeyConfiguration(
        account_key=os.environ.get("AZURE_STORAGE__ACCOUNT_KEY")
    ),
)

mlClient.datastores.create_or_update(datastore)
```
